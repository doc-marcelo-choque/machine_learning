# Prediccion de ventas con regresion lineal simple

Este notebook utiliza el numero de visitantes de una pagina web para predecir su numero de ventas mediante un modelo de regresion lineal simple.

## 1. Importacion de librerias

Se importan las librerias necesarias para manipular los datos, visualizar los resultados y entrenar el modelo.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
import openpyxl
```

## 2. Carga y exploracion de los datos

Se carga el archivo Excel sin modificar el dataset original. Despues se inspeccionan las primeras filas y la estructura de las columnas.

```python
df = pd.read_excel("../data/dataset_comercio_electronico.xlsx")

df.head()
```

```python
df.info()
```

## 3. Estadisticas descriptivas y valores nulos

Se revisan las principales medidas descriptivas y la cantidad de valores faltantes de cada variable.

```python
df.describe()
```

```python
df.isnull().sum()
```

## 4. Relacion entre visitantes y ventas

Se construye un grafico de dispersion para observar visualmente la relacion entre las dos variables.

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["Visitantes"], df["Ventas"], color="steelblue", alpha=0.7)
plt.title("Visitantes y ventas")
plt.xlabel("Numero de visitantes")
plt.ylabel("Numero de ventas")
plt.grid(True, alpha=0.3)
plt.show()
```

## 5. Correlacion

Se calcula la correlacion de Pearson entre visitantes y ventas para medir la intensidad y direccion de su relacion lineal.

```python
correlacion = df["Visitantes"].corr(df["Ventas"])
print(f"Correlacion entre Visitantes y Ventas: {correlacion:.4f}")
```

## 6. Separacion de los datos

Se define X como el numero de visitantes y Y como el numero de ventas. Se reserva el 20% final de los datos para prueba y se conserva el orden cronologico al desactivar la mezcla aleatoria.

```python
X = df[["Visitantes"]]
y = df["Ventas"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    shuffle=False
)

print(f"Registros de entrenamiento: {len(X_train)}")
print(f"Registros de prueba: {len(X_test)}")
```

## 7. Entrenamiento del modelo

Se entrena un modelo de regresion lineal con el 80% inicial de los datos.

```python
modelo = LinearRegression()
modelo.fit(X_train, y_train)
```

## 8. Ecuacion de regresion

Se muestran el coeficiente y el intercepto del modelo. Tambien se grafica la recta estimada sobre los datos de entrenamiento.

```python
coeficiente = modelo.coef_[0]
intercepto = modelo.intercept_

print(f"Ventas = {intercepto:.4f} + ({coeficiente:.4f} * Visitantes)")
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(X_train["Visitantes"], y_train, color="steelblue", alpha=0.6, label="Datos reales")
plt.plot(X_train["Visitantes"], modelo.predict(X_train), color="crimson", label="Recta de regresion")
plt.title("Modelo de regresion lineal")
plt.xlabel("Numero de visitantes")
plt.ylabel("Numero de ventas")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 9. Evaluacion del modelo

Se calculan las metricas MAE, RMSE y R² usando el conjunto de prueba. Estas metricas permiten observar el error de prediccion y la proporcion de variacion explicada por el modelo.

```python
y_pred = modelo.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"MAE:  {mae:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R²:   {r2:.4f}")
```

## 10. Valores reales frente a valores estimados

Se comparan las ventas reales del conjunto de prueba con las ventas estimadas por el modelo.

```python
posiciones = np.arange(len(y_test))

plt.figure(figsize=(10, 5))
plt.plot(posiciones, y_test.to_numpy(), marker="o", label="Ventas reales")
plt.plot(posiciones, y_pred, marker="x", linestyle="--", label="Ventas estimadas")
plt.title("Ventas reales frente a ventas estimadas")
plt.xlabel("Observacion de prueba")
plt.ylabel("Numero de ventas")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 11. Predicciones para nuevos valores

Se estima el numero de ventas para paginas web con 500, 800 y 1200 visitantes.

```python
nuevos_visitantes = pd.DataFrame({"Visitantes": [500, 800, 1200]})
nuevas_ventas = modelo.predict(nuevos_visitantes)

predicciones = pd.DataFrame({
    "Visitantes": nuevos_visitantes["Visitantes"],
    "Ventas estimadas": nuevas_ventas
})

predicciones
```
